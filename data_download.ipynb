```python
import pandas as pd
import tqdm as notebook_tqdm
```

```python
chord_df = pd.read_csv("hf://datasets/ailsntua/Chordonomicon/chordonomicon_v2.csv", low_memory=False)
spotify_track_df = pd.read_csv("hf://datasets/maharshipandya/spotify-tracks-dataset/dataset.csv")
```

```python
# clean up columns
if chord_df.index.name != 'id':
    chord_df.set_index('id', inplace=True)

if "Unnamed: 0" in spotify_track_df.columns.tolist():
    spotify_track_df.drop(columns=["Unnamed: 0"], inplace=True)

spotify_track_df.drop_duplicates(inplace=True)
```

```python
from IPython.display import display

chord_songs = (
    chord_df.reset_index()
    .dropna(subset=["spotify_song_id"])
    .drop_duplicates(subset=["spotify_song_id"], keep="first")
)
spotify_songs = (
    spotify_track_df.dropna(subset=["track_id"])
    .drop_duplicates()
    .drop_duplicates(subset=["track_id"], keep="first")
)
joined = chord_songs.merge(
    spotify_songs,
    left_on="spotify_song_id",
    right_on="track_id",
    how="inner",
)

chord_artists = (
    chord_df.reset_index()
    .dropna(subset=["spotify_artist_id"])
    .drop_duplicates(subset=["spotify_artist_id"], keep="first")
)
spotify_artists = (
    spotify_track_df.dropna(subset=["artists"])
    .drop_duplicates()
    .drop_duplicates(subset=["artists"], keep="first")
)

display(chord_songs.head())
display(joined.head())
display(chord_artists.head())
display(spotify_artists.head())
```
